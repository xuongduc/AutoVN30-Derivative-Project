```python
# library
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
from vnstock import Quote
import csv
```

```python
# constants
data_source = "VCI"  # "TCB", "MSN"
data_interval = "1d"  # "1m", ...
data_startdate = "2000-01-01"
```

```python
# basic functions
def view(obj):
    print(type(obj))
    print(obj)


def to_csv(df, filename="output.csv"):
    if df.empty:
        print("Không có dữ liệu để xuất.")
        return
    
    df.to_csv(filename, index=False)
```

```python
# crawl data
today = datetime.now().date()
quote = Quote(symbol="VN30", source=data_source)
df = quote.history(start=data_startdate, end=str(today), interval=data_interval)
view(df)
to_csv(df, "data.csv")
```
